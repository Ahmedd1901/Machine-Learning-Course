```python
import numpy as np
numbers = np.array([1,2,3,4,5])
result = numbers * 2
print(result)
```

```python
## list vs numpy array
lst = [1,2,3,4]
numarr = np.array([1,2,3,4])
print(lst * 2)
print(numarr * 2)
```

```python
## ndim , shape, size , dtype
arr = np.array([
    [1,2,3],
    [4,5,6]
])

print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
print(arr.itemsize) ## int64 uses 64 bits so 64 / 8 = 8
print(arr.nbytes) ## 8 * 6
```

```python
arr2 = np.zeros(5)
print(arr2)
print(arr2.ndim)
arr3 = np.zeros((2,3))
print(arr3)
print(arr3.ndim)
```

```python
arr4 = np.full((2,3), 10)
print(arr4)
```

```python
arr5 = np.arange(2,10)
print(arr5)
```

```python
arr6 = np.linspace(0,10, 6)
print(arr6)
```

```python
idmat = np.eye(4)
print(idmat)
```

```python
idmat2 = np.identity(3)
print(idmat2)
```

```python
grades = np.array([49,67,88,69,96])
print(grades.min())
print(grades.max())
print(grades.mean())
```
