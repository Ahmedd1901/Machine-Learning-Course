```python
## basic arithmetic
import numpy as np
```

```python
arr = np.array([1,2,3,4,5])
print(arr + 10)
print(arr ** 2)
```

```python
a = np.array([1,2,3])
b = np.array([5,6,7])
print(a + b)
```

```python
## operations and comparisons are element-wise
```

```python
## functions
arr = np.array([1,4,9,16])
res = np.sqrt(arr)
print(arr)
print(type(arr))
print(res)
print(type(res))
```

```python
arr = np.array([0,1,2])
print(np.exp(arr))
```

```python
arr = np.array([1,np.e,np.e**2])
print(np.log(arr))
print(np.log10(arr))
```

```python
arr = np.array([-5,4,3,-2,1])
print(np.abs(arr))
```

```python
arr = np.array([2.445, 3.556,4.564])
print(np.round(arr,2))
```

```python
arr = np.array([2.445, 3.556,4.564,5.6,6.3,8.1])
print(np.floor(arr))
print(np.ceil(arr))
```

```python
## standardization
arr = np.array([10,20,30,40,50])
mean = np.mean(arr)
std = np.std(arr)
z = (arr - mean) / std
print(z)
```
