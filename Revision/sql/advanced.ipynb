```python
import sqlite3
connection = sqlite3.connect('employees2')
cursor = connection.cursor()
```

```python
name = input("Enter name: ")

cursor.execute(f"""
SELECT *
FROM employees
WHERE name='{name}'
""")
print(cursor.fetchall())
## this is a bad way, if the user enters Ahmed' OR 1=1 --
## it will return every row 
##sql injection
```

```python
name = input("Enter name: ")
salary = int(input("Enter salary: "))

cursor.execute(
    '''
    SELECT * FROM employees
    WHERE name = ?
    AND salary > ?
    '''
    , (name, salary)
)
print(cursor.fetchall())
```
