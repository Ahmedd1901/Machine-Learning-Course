```python
import sqlite3
conn = sqlite3.connect('employees2')
cursor = conn.cursor()
```

```python
cursor.execute(
'''
SELECT city, AVG(salary) AS average_salary
FROM employees
GROUP BY city
ORDER BY average_salary
'''   
)
print(cursor.fetchall())
```

```python
cursor.execute(
'''
SELECT city, MAX(salary)
FROM employees
GROUP BY city
'''   
)
print(cursor.fetchall())
```

```python
cursor.execute(
'''
SELECT city,age, MAX(salary)
FROM employees
GROUP BY city,age
'''   
)
print(cursor.fetchall())
```

```python
cursor.execute(
'''
SELECT city, AVG(salary) AS average_salary
FROM employees
where salary > 4000
GROUP BY city
ORDER BY average_salary DESC
LIMIT 2
'''   
)
print(cursor.fetchall())
```

```python
## HAVING
cursor.execute(
'''
SELECT city, AVG(salary) AS average_salary
FROM employees
GROUP BY city
HAVING average_salary > 5000
'''   
)
print(cursor.fetchall())
```

```python
cursor.execute(
'''
SELECT COUNT(*)
FROM employees

'''   
)
print(cursor.fetchall())
```

```python
cursor.execute(
'''
SELECT COUNT(*)
FROM employees
HAVING COUNT(*) > 3 
'''   
)
print(cursor.fetchall())
```
