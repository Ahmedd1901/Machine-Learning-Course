```python
## CREATING TABLE
import sqlite3
conn = sqlite3.connect('employees2')
cursor = conn.cursor()
cursor.execute(
'''
CREATE TABLE IF NOT EXISTS employees(
id INTEGER PRIMARY KEY,
name TEXT,
age INTEGER,
city TEXT,
salary REAL
)

'''
)
```

```python
## INSERTING DATA
cursor.execute(
    '''
    INSERT INTO employees(name,age,city,salary)
    VALUES('ahmed',24,'cairo',5000)    
    '''
)
cursor.execute(
    '''
    INSERT INTO employees(name,age,city,salary)
    VALUES('ali',30,'giza',7000)    
    '''
)
cursor.execute(
    '''
    INSERT INTO employees(name,age,city,salary)
    VALUES('mohamed',22,'cairo',4500)    
    '''
)
cursor.execute(
    '''
    INSERT INTO employees(name,age,city,salary)
    VALUES('mahmoud',27,'alexandria',6000)    
    '''
)
cursor.execute(
    '''
    INSERT INTO employees(name,age,city,salary)
    VALUES('sara',24,'giza',5500)    
    '''
)
```

```python
## save and print data
conn.commit()
cursor.execute(
'''
SELECT * FROM employees
'''
)
print(cursor.fetchall())
```

```python
## ORDER BY
cursor.execute(
'''
SELECT * FROM employees
ORDER BY salary
'''
)
print(cursor.fetchall())
```

```python
cursor.execute(
'''
SELECT * FROM employees
ORDER BY salary DESC
'''
)
print(cursor.fetchall())
```

```python
## LIMIT
import sqlite3
conn = sqlite3.connect('employees2')
cursor = conn.cursor()
cursor.execute(
'''
SELECT * FROM employees
ORDER BY salary DESC
LIMIT 3
'''
)
print(cursor.fetchall())
```

```python
## DISTINCT
cursor.execute(
'''
SELECT DISTINCT city FROM employees
'''    
)
print(cursor.fetchall())
```

```python
## count
cursor.execute(
'''
SELECT COUNT(DISTINCT city) FROM employees
'''    
)
print(cursor.fetchall())
```

```python
## LIKE
cursor.execute(
'''
SELECT * FROM employees
WHERE name LIKE 'a%'
'''    
)
print(cursor.fetchall())
```

```python
cursor.execute(
'''
SELECT * FROM employees
WHERE name LIKE '%med%'
'''    
)
print(cursor.fetchall())
```

```python
## WildCard _
cursor.execute(
'''
SELECT * FROM employees
WHERE name LIKE 'a__'
'''    
)
print(cursor.fetchall())
```

```python
## IN
cursor.execute(
'''
SELECT * FROM employees
WHERE city IN ('cairo','giza')
'''    
)
print(cursor.fetchall())
```

```python
cursor.execute(
'''
SELECT * FROM employees
WHERE city NOT IN ('cairo','giza')
'''    
)
print(cursor.fetchall())
```

```python
## Between
cursor.execute(
'''
SELECT * FROM employees
WHERE age BETWEEN 24 AND 30
ORDER BY age ASC
'''    
)
print(cursor.fetchall())
```
