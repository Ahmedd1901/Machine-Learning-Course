```python
## connect to database
import sqlite3
conn = sqlite3.connect('shopping')
cursor = conn.cursor()
```

```python
## create customers table
cursor.execute(
'''
CREATE TABLE IF NOT EXISTS customers(
customer_id INTEGER PRIMARY KEY,
name TEXT,
city TEXT
)
'''
)
```

```python
## create orders table
cursor.execute(
'''
CREATE TABLE IF NOT EXISTS orders(
order_id INTEGER PRIMARY KEY,
customer_id INTEGER,
product TEXT,
amount REAL
)
'''
)
```

```python
## insert data
customers=[
(1,"Ahmed","Cairo"),
(2,"Sara","Giza"),
(3,"Ali","Alexandria")
]

orders=[
(101,1,"Laptop",1000),
(102,1,"Mouse",20),
(103,2,"Phone",800)
]

cursor.executemany(
"INSERT INTO customers VALUES(?,?,?)",
customers

)

cursor.executemany(
"INSERT INTO orders VALUES(?,?,?,?)",
orders
)
conn.commit()
```

```python
## inner join
cursor.execute("""
SELECT 
c.name,
o.product,
o.amount
FROM customers c
JOIN orders o 
ON c.customer_id = o.customer_id
""")

print(cursor.fetchall())
```

```python
## left join
cursor.execute(
'''
SELECT
c.name,
o.product,
o.amount
FROM customers c
LEFT JOIN
orders o
ON c.customer_id = o.customer_id
'''
)
print(cursor.fetchall())
```
