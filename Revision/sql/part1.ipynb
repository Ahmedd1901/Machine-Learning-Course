```python
import sqlite3
## 1- creating database
conn = sqlite3.connect("school.db")

## 2- create cursor which executes sql
cursor = conn.cursor()

## 3- creating table
cursor.execute(
"""
CREATE TABLE students(
id INTEGER PRIMARY KEY,
name TEXT,
age INTEGER
)
"""
)
```

```python
## 4- inserting data
cursor.execute(
"""
INSERT INTO students(name,age)
VALUES('ahmed',24)
"""
)

## 5- save
conn.commit()
```

```python
## 5- query data
cursor.execute(
"""
SELECT * FROM students

"""
)
## fetch data
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
## 6- close connection
conn.close()
```

```python

conn = sqlite3.connect('employees')
cursor = conn.cursor()
cursor.execute(
"""
CREATE TABLE IF NOT EXISTS employees(
id INTEGER PRIMARY KEY,
name TEXT,
salary REAL
)
"""
)

```

```python
cursor.execute(
"""
INSERT INTO employees(name,salary)
VALUES('ahmed',3000)
"""
)
cursor.execute(
"""
INSERT INTO employees(name,salary)
VALUES('mohamed',2000)
"""
)
conn.commit()
```

```python
cursor.execute(
"""
SELECT * FROM employees
"""
)

print(cursor.fetchall())
```

```python
cursor.execute(
"""
SELECT name FROM employees
"""
)
print(cursor.fetchall())
```

```python
cursor.execute(
"""
SELECT * FROM employees
WHERE salary > 2500
"""
)
print(cursor.fetchall())
```

```python
cursor.execute(
"""
SELECT * FROM employees
WHERE name = 'ahmed'
"""
)
print(cursor.fetchall())
```

```python
conn.close()
```
