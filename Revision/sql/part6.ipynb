```python
import sqlite3
conn = sqlite3.connect('employees2')
cursor = conn.cursor()
```

```python
cursor.execute(
    ''' 
    SELECT * FROM employees
    WHERE salary > (SELECT AVG(salary) FROM employees)

    '''
)
```

```python
print(cursor.fetchall())
```

```python
cursor.execute(
    ''' 
    SELECT AVG(salary) FROM employees

    '''
)
print(cursor.fetchall())
```

```python
cursor.execute("""
DELETE FROM employees
WHERE id NOT IN (
    SELECT MIN(id)
    FROM employees
    GROUP BY name, age
)
""")
conn.commit()
```

```python
## common table expressions (CTEs)
cursor.execute(
    ''' 
    SELECT *
    FROM
    (
        SELECT *
        FROM employees
        WHERE salary > 5000
    ) AS high_salary
    WHERE age > 25;

    '''
)
print(cursor.fetchall())
```

```python
cursor.execute(
    ''' 
    WITH 
    high_salary AS
    (
        SELECT *
        FROM employees
        WHERE salary > 5000
    ),

    young_high_salary_employees AS
    (
    SELECT * FROM high_salary
    WHERE age < 30
    )


    SELECT *
    FROM young_high_salary_employees;

    '''
)
print(cursor.fetchall())
```

```python
cursor.execute(
    '''
    SELECT city, AVG(salary)
    FROM employees
    GROUP BY city
    '''
)
print(cursor.fetchall())
```

```python
cursor.execute(
    ''' 
    SELECT name,city,salary,
    ROW_NUMBER() OVER(ORDER BY salary DESC) AS salary_rank,
    RANK() OVER(ORDER BY salary DESC) AS salary_rank,
    DENSE_RANK() OVER(ORDER BY salary DESC) AS salary_rank
    FROM employees
    '''
)
print(cursor.fetchall())
```
