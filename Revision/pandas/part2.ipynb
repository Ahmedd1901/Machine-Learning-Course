```python
import pandas as pd
```

```python
data = {
    'name' : ['ahmed','mohamed','mahmoud','ali','samir'],
    'age' : [24,22,26,32,30],
    'salary' : [8000,5000,9000,4700,None],
    'department' : ['it','hr','it','finance','hr']
}

df = pd.DataFrame(data)
df.head()
```

```python
df.to_csv('data.csv',index=False)
```

```python
import numpy as np

df['salary_level'] = np.where(
    df['salary'] >= 6000,'High','Low'
)
```

```python
## sorting data
df = df.sort_values(
    by='salary',
    ascending=False
)
```
