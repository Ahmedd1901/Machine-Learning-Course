```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
df.head()
```

```python
def increase_salary(x):
    return x * 1.10

df['salary'] = df['salary'].apply(increase_salary)
print(df)
```

```python
df["salary_after_bonus"] = df["salary"] + df["bonus"]
```

```python
print(df)
```

```python
df['level'] = df['age'].apply(
    lambda x: 'senior' if x>=30 else 'junior'
)
```

```python
df.head()
```

```python
df['name'] = df['name'].str.capitalize()
df['department'] = df['department'].str.upper()
```

```python
df.head()
```

```python
df.to_csv('data.csv')
```
