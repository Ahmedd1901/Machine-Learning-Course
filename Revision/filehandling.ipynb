```python
## create file

with open('example.txt','x') as file:
    file.write("hello, ahmed")
```

```python
## using with open because python automatically closes the file
with open('example.txt','r') as file:
    content = file.read()

print(content)
```

```python
## append
with open('example.txt','a') as file:
    file.write("\nhellooo\nhisssss")
```

```python
## print line by line
with open("example.txt") as file:
    for line in file:
        print(line)
```

```python
## prints lines as a list
with open("example.txt") as file:
    lines = file.readlines()

print(lines)
```

```python
lines = ['mog\n','fasdfd\n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
file = open('example.txt')
content = file.readlines()
print(content)
file.close()
```

```python
with open("example.txt", "r") as file:
    # 1. Read the entire file
    content1 = file.read() 
    print("First Read:", content1)
    
    # 2. Try to read again (returns empty string because cursor is at the end)
    content2 = file.read()
    print("Second Read (Before Seek):", content2) 
    
    # 3. Reset cursor to the beginning
    file.seek(0)
    
    # 4. Read it again successfully
    content3 = file.read()
    print("Third Read (After Seek):", content3)
```
