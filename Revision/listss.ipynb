```python
fruits = ['orange', 'apple', 'banana']
print(fruits[-1])
```

```python
fruits.append('pineapple')
print(fruits)
fruits.insert(1,"kiwi")
```

```python
fruits.remove('pineapple')
print(fruits)
```

```python
fruits.sort()
print(fruits)
```

```python
fruits.reverse()
print(fruits)
```

```python
squares = []
for i in range(10):
    squares.append(i**2)

print(squares)
```

```python
squares=[i**2 for i in range(10)]
print(squares)
```

```python
even_squares = [i**2 for i in range(10) if i%2 == 0]
print(even_squares)
```
