```python
## find the largest number in a list
numbers = [3,5,22,33,87,5,4,3,25,97,34,26]
largest = 0
for i in numbers:
    if i > largest:
        largest = i

print(largest)
print(max(numbers))
```

```python
## remove duplicates
l3 = [2,9,6,4,5,7,3,4,5,2,5]

unique_list = []

for i in l3:
    if i not in unique_list:
        unique_list.append(i)

print(unique_list)
```

```python
## second largest in a list
nums = [1,8,5,99,44,32,45,2,4,9,32,4]
largest = second_largest = 0

for i in nums:
    if i > largest:
        second_largest = largest
        largest = i
    elif i > second_largest and i != largest:
        second_largest = i

print(f"largest is {largest}")
print(f"second_largest is {second_largest}")
```

```python
## count vowels in a list of words
words = ['ahmed', 'hany', 'shaker']
vowels = 0

for word in words:
    for char in word.lower():
        if char in "aeiou":
            vowels +=1 

print(vowels)
```

```python
## flatten a list
nested =[[1,2],[3,4],[5,6]]
flat = [item for sublist in nested for item in sublist]
print(flat)
```

```python
## find frequency of each number in a list and find the most frequent
listOfNumbers = [3,6,4,5,8,9,1,3,2,4,5,1,6,7,8,5,3,2]

## 1- frequency
frequency = {}
for num in listOfNumbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print(frequency)

## 2- most frequent 
count = 0 

for i in frequency:
    if frequency[i] > count:
        count = frequency[i]
        mostfreq = i

print(mostfreq)
```

```python
## most frequent 2
numbers = [1, 2, 2, 3, 1, 2, 4]

freq = {}

for num in numbers:
    freq[num] = freq.get(num, 0) + 1

most_frequent = max(freq, key=freq.get)

print(most_frequent)
```
