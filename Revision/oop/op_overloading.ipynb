```python
## create a vector class and implement all the operator functions

class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __add__(self,other):
        x = self.x + other.x
        y = self.y + other.y
        return Vector(x,y)
    def __sub__(self, other):
        x = self.x - other.x
        y = self.y - other.y
        return Vector(x,y)
    def __mul__(self, other):
        x = self.x * other.x
        y = self.y * other.y
        return Vector(x,y)
    def __str__(self):
        return f"Vector({self.x},{self.y})"
    
v1 = Vector(3,4)
v2 = Vector(2,4)
print(v1 + v2)

    
```
