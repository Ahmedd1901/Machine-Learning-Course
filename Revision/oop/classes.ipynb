```python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def introduce(self):
        return f"my name is {self.name}"
## self refers to the current object
student1 = Student('ahmed',24)
print(student1.name)
print(student1.age)
print(student1.introduce())
```

```python
class dataset:
    def __init__(self,name,rows):
        self.name = name
        self.rows = rows
    def summary(self):
        return f"{self.name}: {self.rows} rows"
    
iris = dataset("iris", 150)
print(iris.summary())
```

```python
## class methods
class Student:
    school = 'ain shams'
    def __init__(self):
        pass

    @classmethod
    def get_school(cls):
        return cls.school
    
    @classmethod
    def change_school(cls,new_name):
        cls.school = new_name
    
    
print(Student.get_school())
Student.change_school("cairo university")
print(Student.get_school())
```

```python
## factory method
class Person:

    def __init__(self,name,age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls,data):

        name,age = data.split(",")

        return cls(name,int(age))

p = Person.from_string("Ahmed,24")

print(p.name)
print(p.age)
```

```python
## static method: doesn't need self or cls
class MathUtils:

    @staticmethod
    def add(a,b):
        return a+b
    
print(MathUtils.add(2,3))
```
