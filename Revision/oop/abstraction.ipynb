```python
from abc import ABC,abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
## It forces child classes to implement the method.
## You cannot create abstract objects.
```

```python
from abc import ABC,abstractmethod

class MLModel(ABC):

    @abstractmethod
    def fit(self):
        pass
    @abstractmethod
    def predict(self):
        pass

class DecisionTree(MLModel):

    def fit(self):
        print("Training tree")

    def predict(self):
        print("Predicting tree")
```
