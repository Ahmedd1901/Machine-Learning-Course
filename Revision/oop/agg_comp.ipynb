```python
## Inheritance: Is-a ex. dog is an animal
## composition: Has-a ex. car has an engine
## aggregation: Has-a ex. department has a teacher but
## teacher can exist without department
```

```python
## Composition
class Engine:
    def start(self):
        return "Engine Started"
class Car:
    def __init__(self):
        self.engine = Engine()
    
    def start_car(self):
        return self.engine.start()

car = Car()
print(car.start_car())
```

```python
## aggregation
class Book:
    def __init__(self,name):
        self.name = name

class Library:
    def __init__(self,books):
        self.books = books
    def __str__(self):
        return f"Library contains: {[book.name for book in self.books]}"
    def __repr__(self):
        return f"Library contains: {[book.name for book in self.books]}"  

b1 = Book("book1")
b2 = Book("Book2")
l1 = Library([b1,b2])
l1
```

```python
## ml style
class DataLoader:
    def load(self):
        return "Data loaded"
class Model:
    def train(self):
        return "Model trained"

class MLPipeline:
    def __init__(self):
        self.load = DataLoader()
        self.train = Model()
    def run(self):
        a = self.load.load()
        b = self.train.train()
        return a,b

m1 = MLPipeline()
m1.run()
```
