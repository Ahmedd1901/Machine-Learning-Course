```python
## __init__
class Student:
    def __init__(self,name):
        self.name = name
    
    def __str__(self):
        return f"{self.name}"

s1 = Student("ahmed")
## python automatically calls:
Student.__init__(s1,"ahmed")
```

```python
## str(): how object appear when printed
print(s1)
```

```python
## len(): made it return more than the length by 1
class Playlist:

    def __init__(self,songs):
        self.songs = songs

    def __len__(self):
        count =0
        for i in self.songs:
            count+=1
        
        return count + 1
    
p = Playlist(["song1","song2","song3","song4"])

print(len(p)) 
print(p.__len__())
```

```python
## getitem()
class Dataset:

    def __init__(self,data):
        self.data = data

    def __getitem__(self,index):
        return self.data[index]

dataset = Dataset([10,20,30])

print(dataset[1])
print(dataset.__getitem__(1))
```

```python
class Dataset:

    def __init__(self,data):
        self.data = data

    def __setitem__(self,index,value):
        self.data[index] = value

dataset = Dataset([1,2,3])

dataset[0] = 100

print(dataset.data)
```

```python
class Numbers:

    def __init__(self):
        self.data = [1,2,3]

    def __iter__(self):
        return iter(self.data) 
    
nums = Numbers()

for x in nums:
    print(x)   
```

```python
## call(): object behaves like a function
class Multiplier:

    def __init__(self,factor):
        self.factor = factor

    def __call__(self,x):
        return x * self.factor
    
double = Multiplier(2)

print(double(5))
```
