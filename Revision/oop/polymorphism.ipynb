```python
class Shape:

    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self,w,h):
        self.width = w
        self.height = h
    def area(self):
        return self.height * self.width
    
class Circle(Shape):
    def __init__(self,r):
        self.r = r
    def area(self):
        return (self.r ** 2) * 3.14
    
shapes = [
    Rectangle(5,4),
    Circle(3)
]

for shape in shapes:
    print(shape.area())
    
```
