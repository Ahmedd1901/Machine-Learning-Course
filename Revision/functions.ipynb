```python
## recursion

def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)

print(factorial(5))
```

```python
## find the sum of numbers
def sum_numbers(*args):
    sum = 0
    for i in args:
        sum+= i
    
    return sum

print(sum_numbers(1,2,3,4,5))
```

```python
## find the largest number

def largest(*args):

    largest = args[0]
    for num in args:
        if num > largest:
            largest = num
    
    return largest

print(largest(10,5,30,20))
```

```python
## count vowels in a text

def vowels(text):
    vowels = 'aeiou'
    count = 0
    for i in text.lower():
        if i in vowels:
            count+=1
    
    return count

print(vowels('ahmed'))
```

```python
## price in cart
cart = [
    {"price":10, "quantity":2},
    {"price":5, "quantity":3}
]


def total_cost(cart):
    total = 0

    for item in cart:
        total += item['price'] * item['quantity']

    return total

total_cost(cart)
```

```python
## palindrome check
def ispal(text):
    text = text.lower().replace(" ","")
    return text == text[::-1]

print(ispal("TRoort"))
print(ispal("ahmed"))
```

```python
## count words 
def count_words(sentence):
    words = sentence.split()
    return len(words)

print(count_words("ahmed hany shaker"))
```

```python
## character frequency

def char_freq(text):
    freq = {}
    text = text.lower().replace(" ","")
    for char in text:
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1

    return freq

text1 = "are to he she if"
print(char_freq(text1))
```

```python
## fibonacci 
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

fibonacci(6)
```

```python
## average

def avg(*args):
    """returns the average of given numbers
    """
    sum =0
    for i in args:
        sum+=i
    return sum/len(args)

avg(10,20,30)
```
