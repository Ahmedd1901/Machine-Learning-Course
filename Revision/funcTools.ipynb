```python
## lambda

add = lambda x,y : x + y

add(2,3)
```

```python
even = lambda x : x % 2 ==0 
even(4)
```

```python
## map
nums = [1,2,3,4]
print(list(map(even,nums)))
```

```python
res = list(map(lambda x:x**2,nums))
print(res)
```

```python
words = ["ahmed","mohamed"]
upper = list(map(str.upper,words))
print(upper)
```

```python
## filter
lst = [1,2,3,4,5,6,7,8]

result = list(filter(even,lst))
print(result)
```

```python
result2 = list(filter(lambda x: x > 5,lst))
print(result2)
```

```python
people = [
    {'name':'Ahmed','age':24},
    {'name':'Mohamed','age':20},
    {'name':'Mahmoud','age':26}
]


elig = list(filter(lambda x:x['age'] > 21, people))
print(elig)
```
