```python
scores = {
    "Ahmed": 90,
    "Ali": 75,
    "Sara": 95,
    "Omar": 80
}


max_name = ""
max_score = 0

for name, score in scores.items():
    if score > max_score:
        max_score = score
        max_name = name

print(max_name)
print(max_score)
```

```python
## merge 2 dictionaries
a = {'a' : 1}
b = {'b' : 2}

merged = {**a,**b}
print(merged)
```

```python
## find duplicates in a dictionary

numbers = [1,2,2,3,4,5,5,6,6,6]

frequency = {}
duplicates = []

for num in numbers:
    if num in frequency:
        frequency[num] +=1
    else:
        frequency[num] = 1

print(frequency)

for key,value in frequency.items():
    if value > 1:
        duplicates.append(key)

print(duplicates)
```
