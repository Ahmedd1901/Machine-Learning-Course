```python
## try-except
try:
    x = 10/0
except:
    print("error occurred")
```

```python
## catch specific exceptions
try:
    num = int(input("enter a number: "))
except ValueError:
    print("enter a valid integer")
```

```python
## multiple exceptions
try:
    num = int(input("enter a number"))
    x = 10 / num
except ValueError:
    print("enter a valid integer")
except ZeroDivisionError:
    print("cannot divide by zero")
```

```python
## exception object
try:
    x = int('abc')
except ValueError as e:
    print(e)
```

```python
## else-block runs when no exception occurs
try:
    x = int(input("enter a number"))
except ValueError:
    print("enter a valid number")
else:
    print(f"you entered {x}")
```

```python
## finally-block runs no matter what
try:
    x = int(input("enter a number"))
    val = 10 / x
except ValueError:
    print("enter a valid number")
except ZeroDivisionError:
    print("cannot divide by zero")
else:
    print(val)
finally:
    print("program finished")
```

```python
## index error
lst = [1,2,3]
try:
    lst[10]
except IndexError:
    print("index error")
```

```python
## key error
student = {'name' : 'ahmed'}
try:
    student['age']
except KeyError:
    print("key not found")
```

```python
## file not found
try:
    open('missing.txt')
except FileNotFoundError:
    print("file not found")
```

```python
## type error
try:
    '5' + 5
except TypeError:
    print("type error")
```

```python
## raising exceptions
age = int(input("enter your age"))
if age < 0 :
    raise ValueError('age cannot be negative')
```

```python
scores = []

with open('scores.txt','r') as file:
    for line in file:
        try:
            scores.append(int(line.strip()))
        except ValueError:
            print("invalid data: ",line.strip())

print(scores)
```
