```python
## list inside a tuple can change

t2 = (1,[2,3],5)
t2[1].append(4)
print(t2)
```

```python
a = (1)
print(type(a))
a = (1,)
print(type(a))
```

```python
## sum of even numbers in a tuple

tup = (3,2,4,5,6,3,7,4,8)
sum = 0
for num in tup:
    if num % 2 == 0:
        sum += num

print(sum)
```

```python
## merging tuples
tup1 = (1,2,3)
tup2 = (3,4,5,6)
res = tup1 + tup2
print(res)
```

```python
## flatten a nested tuple
nested = ((1,2),(3,4),(5,6))
flat = []

for subtup in nested:
    for item in subtup:
        flat.append(item)

flat = tuple(flat)
print(flat)
print(type(flat))
```
