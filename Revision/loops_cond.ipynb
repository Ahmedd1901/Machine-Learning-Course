```python
## prime number checking
number = int(input("Enter a number"))
isPrime = True

if number < 2:
    isPrime = False
else:
    for i in range(2,number):
        if number % i == 0:
            isPrime = False
            break
        

if isPrime:
    print(f"{number} is prime")
else:
    print(f"{number} is not prime")

## 2 is prime since the loop body never executes
```

```python
## palindrome checking
text = input("enter a word: ")
ispalindrome = True

for i in range(len(text) // 2):
    if text[i] != text[len(text) - i - 1]:
        ispalindrome = False
        break

if ispalindrome:
    print(f"{text} is palindrome")
else:
    print(f"{text} is not palindrome")
```

```python
## fibonacci sequence
number = int(input("enter a number"))
a = 0
b = 1
for i in range(number):
    print(a)
    a,b = b,b+a
```

```python
## factorial
number = int(input("enter a number"))
factorial = 1
i = 1
while i <= number:
    factorial *= i
    i += 1

print(factorial)

```

```python
## sum of digits
number = int(input("enter a number"))
saved_number = number
sum_digits = 0

while number > 0:
    digit = number % 10
    sum_digits += digit
    number //= 10

print(f"The total number of digits in {saved_number} is {sum_digits}")
```

```python
## Login validation
correct_username = 'admin'
correct_password ='1234'
attempts = 3

while attempts > 0:
    username = input("enter username")
    password = input("enter password")

    if username == correct_username and password == correct_password:
        print("successful login")
        break
    attempts -= 1
    print('invalid credentials')

if attempts == 0:
    print('account locked')
```

```python
## password validation
password = input("enter a password")
hasupper = False
hasdigit = False
haslower = False

for char in password:
    if char.isupper():
        hasupper = True
    if char.isdigit():
        hasdigit = True
    if char.islower():
        haslower = True
if len(password) >= 8 and hasdigit and hasupper and haslower:
    print("valid")
else:
    print("invalid")
```

```python
## ATM
balance = 10000

while True:
    print("1- check balance")
    print("2- Deposit")
    print("3- withdraw")
    print("4- Exit")

    choice = input("choose: ")

    if choice == "1":
        print(f'Balance: {balance}')
    elif choice == "2":
        amount = float(input("enter amount"))
        balance += amount
        print(f"balance: {balance}")
    elif choice == "3":
        amount = float(input("enter amount"))
        if amount <= balance:
            balance -= amount
            print(f"balance: {balance}")
        else:
            print("insufficient amount")
    elif choice == "4":
        print("goodbye")
        break
    else:
        print("invalid choice")

```

```python
## finding max and min
numbers = [8,2,5,10,3,7]
maximum = numbers[0]
minimum = numbers[0]

for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

print(f"maximum: {maximum}")
print(f"minimum: {minimum}")
```

```python
## frequency of characters
text = "ahmed hany"
frequency = {}

for char in text:
    if char != " ":
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

print(frequency)
```

```python
## frequency of each vowel and total number of vowels

text =" I went to the zoo yesterday"
vowels = 0 
frequency = {}
for char in text.lower():
    if char in "aeiou":
        vowels += 1
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

print(f"There are {vowels} vowels")
print(frequency)
```

```python
## finding the second largest number in a list
list = [10,13,5,8,20,22,13,19]
largest = second_largest = 0

for num in list:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print(f"largest is {largest}")
print(f"second_largest is {second_largest}")
```

```python
## find the most frequent char in a string
text = input("enter a string")
frequency = {}
for char in text:
    if char != ' ':
        if char in frequency:
            frequency[char] +=1 
        else:
            frequency[char] = 1

most_freq = ''
max_count = 0

for i in frequency:
    if frequency[i] > max_count:
        max_count = frequency[i]
        most_freq = i
    
print(f"{most_freq}: {max_count}")
```
