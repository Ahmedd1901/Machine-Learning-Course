```python
## FizzBuzz
for num in range(1,21):
    if num % 3 == 0 and num % 5 == 0:
        print("FizzBuzz")
    elif num % 3 == 0:
        print("Fizz")
    elif num % 5 == 0:
        print("Buzz")

    
```

```python
## palindrome
word = input("enter a word")
isPal = True

for char in range(len(word)//2):
    if word[char] != word[len(word) - char - 1]:
        isPal = False
        break

print(isPal)
```

```python
## find the second largest number
numbers = [10,25,3,50,20]
largest = second = 0
for num in numbers:
    if num > largest:
        second = largest
        largest = num
    elif num > second and num != largest:
        second = num
    
print(largest)
print(second)
```

```python
## frequency counter
text = "python java python c++ java python"
freq = {}
words = text.split()
for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print(freq)
```

```python
## remove duplicates
l1 = [1,2,2,3,4,4,5]
unique_list = []

for num in l1:
    if num not in unique_list:
        unique_list.append(num)

print(unique_list)
```

```python
## student grades
grades = {
    "Ahmed":90,
    "Ali":75,
    "Sara":95,
    "Omar":80
}
highest = float('-inf')
lowest = float('+inf')

for key,value in grades.items():
    if value > highest:
        highest = value
        best = key

for key,value in grades.items(): 
    if value < lowest:
        lowest = value
        worst = key

print(highest,best)
print(lowest,worst)

sum = 0
count = 0
for val in grades.values():
    sum += val
    count += 1
print(sum/count)
```
