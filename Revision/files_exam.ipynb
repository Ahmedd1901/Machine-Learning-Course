```python
## create a function that returns the number of lines in a file

def countlines(file):
    with open(file,'r') as file:
        return len(file.readlines())
    
countlines('example.txt')
```

```python
## count words

def count_words(file):
    with open(file,'r') as file:
        words = file.read().split()
        return len(words)
    
count_words('example.txt')
```

```python
## find longest word

def longest_word(file):
    with open(file,'r') as file:
        words = file.read().split()
        return max(words,key=len)
    
longest_word('example.txt')
```

```python
## copy file
def copy(src,dest):
    with open(src,'r') as src:
        content = src.read()
    
    with open(dest,'w') as dest:
        dest.write(content)

copy('example.txt','example2.txt')
```

```python
## count character frequency

def char_freq(file):
    freq = {}
    with open(file,'r') as file:
        content = file.read()
        for char in content.lower():
            if char.isalpha() or char.isdigit():
                freq[char] = freq.get(char,0) + 1
        
    return freq

char_freq('example.txt')
```

```python
scores = []
with open('scores.txt','r') as file:
    for line in file:
        scores.append(int(line.strip()))

print(scores)
avg = sum(scores) / len(scores)
max = max(scores)
min = min(scores)
print(avg,min,max)
```
