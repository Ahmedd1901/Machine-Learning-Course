```python
def decorator(func):
    def wrapper(*args,**kwargs):
        result = func(*args,**kwargs)
        return f"before\n{result}\nafter"
    return wrapper

def hello():
    return "hello"

print(decorator(hello)())
```

```python
@decorator
def add(a,b):
    return a+b

print(add(2,3))
```

```python
import time

def timer(func):

    def wrapper(*args, **kwargs):

        start = time.time()

        result = func(*args, **kwargs)

        end = time.time()

        print(f"Execution time: {end-start}")

        return result

    return wrapper

@timer
def slow_function():

    time.sleep(2)

print(slow_function())
```
