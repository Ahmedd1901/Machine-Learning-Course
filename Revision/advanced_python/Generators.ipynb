```python
def my_generator():

    yield 1
    yield 2
    yield 3

gen = my_generator()

print(next(gen))
```

```python
def numbers():

    print("Start")

    yield 1

    print("Middle")

    yield 2

    print("End")

    yield 3
```

```python
gen = numbers()
next(gen)
```

```python
next(gen)
```

```python
def fibonacci(n):

    a,b = 0,1

    for _ in range(n):

        yield a

        a,b = b,a+b
for num in fibonacci(10):
    print(num)
```

```python
def infinite_counter():

    num = 1

    while True:

        yield num

        num += 1
counter = infinite_counter()

print(next(counter))
print(next(counter))
print(next(counter))
```

```python
next(counter)
```
